## 4. The data toolkit

`numpy`, `pandas`, `matplotlib`, and `scipy` are pre-installed — the libraries you'll meet in virtually every data / scientific Python tutorial. A 30-second taste:

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "day": pd.date_range("2026-01-01", periods=14),
    "minutes_practiced": np.random.default_rng(42).integers(10, 90, 14),
})
df.head()
```

```python
import matplotlib.pyplot as plt

ax = df.plot(x="day", y="minutes_practiced", marker="o", legend=False)
ax.set_title("Practice minutes per day")
ax.set_ylabel("minutes")
plt.show()
```

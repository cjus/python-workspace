## 1. Check your environment

This confirms the notebook is using the workspace's own private Python setup and that the libraries are available. If anything shows ❌, run the launcher again (`./start.sh` — Windows: `start.cmd`; it installs whatever is missing), then pick **Kernel → Restart Kernel** from the menu and re-run this cell.

```python
import sys

print(f"Python {sys.version.split()[0]}")
print(f"Interpreter: {sys.executable}")

for name in ("numpy", "pandas", "matplotlib", "scipy"):
    try:
        mod = __import__(name)
        print(f"✅ {name} {mod.__version__}")
    except ImportError:
        print(f"❌ {name} — not installed (run ./start.sh again)")
```

## 2. AI in the chat panel

Open a chat from the JupyterLab launcher (the **Chat** card) or the chat icon in the left sidebar, then `@`-mention a persona:

- **`@Jupyternaut`** — the general assistant. Ask it anything: *“explain this error”*, *“what does this cell do?”*. It can read/write files and work with notebooks (it asks permission first).
- **`@Tutor`** — the built-in Python tutor (see [`.jupyter/personas/tutor_persona.py`](.jupyter/personas/tutor_persona.py)). It answers learning questions **and** scaffolds practice notebooks:

The **first time**, pick your model in **Settings → Jupyternaut Settings** — e.g. `ollama/gemma4:26b-mlx` on Apple Silicon, `ollama/gemma4:12b` on Windows/Linux (`-mlx` builds are Apple-only), or an `openrouter/...` model for hosted. All personas and the `%%ai` magics use this selection.

## 3. AI inside cells — the `%%ai` magics

You can also ask the AI right inside a notebook. Two bits of one-time setup per session: load the extension, then give your local model a short nickname (an **alias**) — after that, `%%ai <nickname>` asks the AI anything. (Cloud models like `openrouter/...` can be used directly, no alias needed.)

```python
%load_ext jupyter_ai_magic_commands
```

```python
# Point the alias at any model from `ollama list`.
# macOS (Apple Silicon) repo default:
%ai alias gemma ollama/gemma4:26b-mlx
# Windows / Linux: `-mlx` builds are Apple-only — comment the line above and
# use the standard tag instead:
# %ai alias gemma ollama/gemma4:12b
```

```python
%%ai gemma
In two short paragraphs: what is a Python list, and when would I use a tuple instead?
```

```python
%%ai gemma -f code
Write a Python function that returns the nth Fibonacci number, with a docstring.
```

## 4. The data toolkit

`numpy`, `pandas`, `matplotlib`, and `scipy` are pre-installed — the libraries you'll meet in virtually every data / scientific Python tutorial. A 30-second taste:

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "day": pd.date_range("2026-01-01", periods=14),
    "minutes_practiced": np.random.default_rng(42).integers(10, 90, 14),
})
df.head()
```

```python
import matplotlib.pyplot as plt

ax = df.plot(x="day", y="minutes_practiced", marker="o", legend=False)
ax.set_title("Practice minutes per day")
ax.set_ylabel("minutes")
plt.show()
```

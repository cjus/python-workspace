## 1. Errors are not failures (~3 min)

Last week you wrote your own **functions** with `def` and `return`. The more code you write, the more often Python will stop and show you a red **error message**. Here's the secret nobody tells beginners: 🤫

**Every programmer sees errors all day — including professionals.** Errors are not a sign you're bad at this. They're just Python trying to *help* you, in the only way it can: by stopping and pointing at the problem.

So today we flip the script. Instead of fearing that red text, we'll learn to **read it, fix it, and even expect it**. By the end, an error will feel less like a slammed door and more like a sticky note that says "hey, look here." ✅

## 2. Reading a traceback (~10 min)

When Python hits a problem it prints a **traceback** (also called a "stack trace"). It can look like a wall of text, but there's a simple trick:

> 👉 **Read the LAST line first.** It names the **type** of error and gives a short **message**.

Above that last line, Python shows the **line of code** that broke, often with a little `^^^^^` **caret** marker pointing right at the trouble spot. Modern Python is friendly: it even adds *"Did you mean…?"* suggestions when it can guess what you meant.

Let's tour the handful of errors you've almost certainly already bumped into. For each one, the broken code is **commented out** (so the cell still runs), and the real traceback is shown right below in a grey box.

*(Note: your traceback may look slightly different — paths and exact wording vary by computer and Python version — but the shape is always the same.)*

### NameError — a typo in a name

A **NameError** means you used a name Python has never heard of — usually a **typo** in a variable or function name.

````python
# greeting = "Hello!"
# print(greting)        # oops: we wrote "greting", not "greeting"

print("The broken lines above are commented out so this cell runs. ✅")
````

```
Traceback (most recent call last):
  Cell In[1], line 2, in <module>
    print(greting)
          ^^^^^^^
NameError: name 'greting' is not defined. Did you mean: 'greeting'?
```

Read the last line: **NameError**, name `'greting'` is not defined. The caret points at `greting`, and Python even guesses **"Did you mean: 'greeting'?"** — that hint is gold. 💡

### TypeError — mixing types that don't go together

A **TypeError** means you tried to do something with the *wrong kind* of value — like adding text to a number, or calling a function the wrong way.

````python
# age = 1
# print("You are " + age + " years old")   # can't glue text and a number with +

print("Commented out so this cell runs. ✅")
````

```
Traceback (most recent call last):
  Cell In[2], line 2, in <module>
    print("You are " + age + " years old")
          ~~~~~~~~~~~^~~~~
TypeError: can only concatenate str (not "int") to str
```

The message spells it out: you can only `+` a **str** (text) onto another **str**, not an **int** (whole number). The fix is to convert with `str(age)`, or use an f-string: `f"You are {age} years old"`.

A TypeError also shows up when you call a function the wrong way. For example, `len()` needs *one* thing to measure:

````python
# print(len())     # len needs something to measure!

print("Commented out so this cell runs. ✅")
````

```
Traceback (most recent call last):
  Cell In[3], line 1, in <module>
    print(len())
          ~~~^^
TypeError: len() takes exactly one argument (0 given)
```

"takes exactly one argument (0 given)" → we forgot to put something inside the parentheses, like `len("hello")`.

### IndexError — reaching past the end of a list

Remember from Lesson 4 that list positions (**indexes**) start at `0`. An **IndexError** means you asked for a position that doesn't exist.

````python
colors = ["red", "green", "blue"]   # valid indexes are 0, 1, 2

# print(colors[5])   # there is no item at position 5!

print("There are", len(colors), "colors, so the last index is", len(colors) - 1)
````

```
Traceback (most recent call last):
  Cell In[4], line 3, in <module>
    print(colors[5])
          ~~~~~~^^^
IndexError: list index out of range
```

**IndexError: list index out of range** — the list only has 3 items (indexes 0, 1, 2), so `colors[5]` is off the edge.

### SyntaxError — a typo in the grammar

A **SyntaxError** is different from the others: Python can't even *start* running because the code isn't written correctly — a missing colon `:`, an unclosed parenthesis, that sort of thing. It usually points at the line *where it noticed* the problem.

````python
# if 3 > 2          # missing the colon at the end!
#     print("yes")

print("Commented out so this cell runs. ✅")
````

```
  Cell In[5], line 1
    if 3 > 2
            ^
SyntaxError: expected ':'
```

**SyntaxError: expected ':'** — and the caret sits right where the colon should go. Notice there's no "Traceback" header here: the code never ran at all, so there's nothing to trace. 🧱

## 3. Detective work: debugging (~9 min)

Some bugs are sneakier than a traceback. The code **runs with no error** — but the answer is just *wrong*. 😬 There's no red text to read, so you have to do some detective work. This is called **debugging**.

Your number-one tool costs nothing and you already know it: **`print()`**. Sprinkle a few prints inside your code to *see what it's actually doing*, step by step. The computer can't lie to you — print reveals the truth.

Here's a function that's *supposed* to return the average of a list of numbers. The working version first:

````python
def average(numbers):
    """Return the average (mean) of a list of numbers."""
    total = sum(numbers)          # sum() is a built-in that adds up a list for us
    count = len(numbers)          # how many there are
    return total / count          # divide to get the average

scores = [10, 20, 30, 40]
print("Average:", average(scores))   # (10+20+30+40) / 4 = 25.0
````

## 4. try / except: errors you expect (~12 min)

Here's a key idea: **not every error is a bug.** Some errors are just *normal life*. A user types `"forty-two"` when you wanted a number. A file you're looking for isn't there (that's next week 👀). Your program shouldn't *crash* over these — it should handle them gracefully.

For example, turning text into a number with `int()` works great… until the text isn't a number:

````python
# number = int("forty-two")   # this raises a ValueError

print("Commented out so this cell runs. ✅")
````

```
Traceback (most recent call last):
  Cell In[9], line 1, in <module>
    number = int("forty-two")
ValueError: invalid literal for int() with base 10: 'forty-two'
```

That's a **ValueError** — the value isn't something `int()` can handle. We can *catch* errors like this with **`try` / `except`**:

```
try:
    risky_thing()            # code that MIGHT fail
except ValueError:
    handle_it()              # runs ONLY if a ValueError happened
```

Python runs the `try` block; if an error of that type pops up, it jumps to the `except` block instead of crashing. Let's see it.

````python
try:
    number = int("forty-two")
    print("Converted:", number)
except ValueError:
    print("Hmm, that wasn't a number I could convert. 🙃")

print("...and the program keeps going! ✅")
````

Now something realistic: a list of messy text values, some of which aren't numbers. We try to convert each one, and give a **friendly message** for the ones that fail — instead of the whole program blowing up on the first bad value.

````python
messy = ["42", "seven", "100", ""]   # a preset list — no typing needed

for item in messy:
    try:
        value = int(item)
        print(f"'{item}' -> {value} ✅")
    except ValueError:
        print(f"'{item}' is not a whole number, skipping. ⏭️")
````

### Catch *specific* errors

Notice we wrote `except ValueError:` — naming the exact error we expect. You *can* write a bare `except:` that catches **everything**, but that's a **bad habit**: it hides bugs you didn't expect (like a typo'd variable name) by silently swallowing them. Catch the specific error you're prepared to handle, and let surprises surface so you can fix them. 🎯

This pattern — try the risky thing, handle the expected failure kindly — is how real programs stay **friendly instead of crashing**.

## 5. Payoff: a crash-proof program (~3 min)

We've spent the lesson staring at red error text — so let's end with what it's all *for*. Below is a tiny program that loops over a list of deliberately **bad** inputs and survives every single one, using the very `try` / `except` patterns you just learned. Watch it shrug off the junk and keep going. 💪

````python
# A preset list of deliberately messy inputs — no typing needed.
bad_inputs = ["12", "banana", "", "-5", "3.7"]

survived = 0
for raw in bad_inputs:
    try:
        number = int(raw)            # the risky step: text -> whole number
        if number > 0:
            verdict = "positive 👍"
        else:
            verdict = "zero or negative 👀"
        print(f"'{raw}' -> {number}, that's {verdict}")
    except ValueError:
        # catches both "banana"/"" (not a number) AND "3.7" (not a WHOLE number)
        print(f"'{raw}' -> not a whole number, skipped it safely. ⏭️")
    survived += 1                    # we got through this input without crashing

print(f"\n✅ survived all {survived} bad inputs without crashing — THIS is why errors are not failures.")
````

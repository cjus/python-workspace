## 2. Variables (~10 min)

A **variable** is a name you give to a piece of information so you can use it later. Think of it as a **labeled box** (or a sticky note): you put something in, write a name on the front, and reach for it whenever you need it.

You create one with `=` (the **assignment** sign). It means *"put the thing on the right into the name on the left."* It does **not** mean "equals" like in math.

````python
age = 30          # a number goes into the box named "age"
name = "Sam"      # text goes into the box named "name"

print(age)
print(name)
````

**Naming your boxes** 🏷️ — a few simple rules and one strong habit:

- Use lowercase words joined by underscores: `first_name`, `total_cost`. This style is called **snake_case** (it looks like a snake lying down 🐍).
- Pick **descriptive** names: `score` beats `x`.
- A name **can't start with a digit** (`2cats` is not allowed, but `cats2` is fine).

You can also **reassign** a variable — just put something new in the box. The old value is gone.

````python
score = 10
print(score)

score = 25        # reassign: the box now holds 25
print(score)

# You can also use a variable inside an expression:
score = score + 5  # take what's in score, add 5, put it back
print(score)
````

## 3. Numbers (~8 min)

Python has two kinds of numbers you'll meet constantly:

- **`int`** — a whole number, like `3` or `-12` (short for *integer*).
- **`float`** — a number with a decimal point, like `3.5` or `0.1` (short for *floating-point*).

The everyday math operators are what you'd expect: `+` add, `-` subtract, `*` multiply, `/` divide.

````python
print(2 + 3)     # 5
print(10 - 4)    # 6
print(6 * 7)     # 42
print(10 / 4)    # 2.5  — division always gives a float
````

Three more handy operators:

- `//` — **floor division**: divide and throw away the remainder (whole pizzas only).
- `%` — **modulo**: just the remainder (leftover slices).
- `**` — **power**: `2 ** 3` means 2 to the power of 3.

🍕 Imagine 17 pizza slices shared among 5 friends: `//` tells you how many *whole* slices each gets, and `%` tells you how many are left over.

````python
slices = 17
friends = 5

print(slices // friends)   # 3  -> each friend gets 3 whole slices
print(slices % friends)    # 2  -> 2 slices left over
print(2 ** 3)              # 8  -> 2 * 2 * 2

# Order of operations is like math class: * and / happen before + and -
print(2 + 3 * 4)           # 14, not 20
print((2 + 3) * 4)         # 20 -> parentheses go first
````

## 4. Text — strings (~10 min)

Text in Python is called a **string** (a *string* of characters). You write it inside quotes — single `'...'` or double `"..."`, your choice, as long as they match.

You can join strings together with `+`. This is called **concatenation**.

````python
greeting = "Hello"
who = 'world'

print(greeting + ", " + who + "!")   # Hello, world!
````

### The star of the show: f-strings ⭐

Joining with `+` gets clumsy fast. **f-strings** are a much nicer way to drop variables into text. Put the letter `f` right before the opening quote, then write `{variable}` anywhere inside — Python fills in the value for you.

````python
name = "Sam"
age = 30

# Without f-string (clumsy, and note: numbers need care to join — more on that soon)
print("Hi " + name + "!")

# With an f-string (clean and clear):
print(f"Hi {name}, you are {age} years old.")

# You can even do math inside the curly braces:
print(f"Next year you'll be {age + 1}.")
````

Strings come with built-in **methods** — little actions you call with a dot. A few favorites (plus two non-method extras worth knowing alongside them):

- `.upper()` / `.lower()` — change the case.
- `.title()` — Capitalize Each Word.
- `len(text)` — how many characters. This one is a built-in *function*, not a dot-method, so it wraps the value: `len(text)`, no dot.
- `"ha" * 3` — `*` repeats a string: `"hahaha"`. That's the multiply *operator*, not a method either — but handy to know.

````python
word = "python"

print(len(word))        # 6
print(word.upper())     # PYTHON
print(word.title())     # Python
print("ha" * 3)         # hahaha
````

## 5. Types & conversions (~8 min)

Every value in Python has a **type** — is it a number? text? You can ask with `type(...)`.

````python
print(type(7))        # <class 'int'>
print(type(3.5))      # <class 'float'>
print(type("hello"))  # <class 'str'>
print(type("7"))      # <class 'str'>  -> quotes make it TEXT, not a number!
````

### The classic beginner crash 💥

This is the mistake *everyone* makes once. Look closely:

```python
"1" + 1
```

`"1"` (with quotes) is **text**, and `1` is a **number**. Python doesn't know whether you mean to glue text together or add numbers, so it stops and complains. The fix is to make both sides the same type — convert one to match the other:

- `int("1")` turns the text `"1"` into the number `1`.
- `float("3.5")` turns text into a decimal number.
- `str(1)` turns the number `1` into the text `"1"`.

````python
# Convert text to a number, THEN add:
print(int("1") + 1)        # 2  -> both are numbers now

# Convert a number to text, THEN join:
print("Score: " + str(1))  # Score: 1  -> both are text now

# Or skip the hassle entirely with an f-string:
print(f"Score: {1}")       # Score: 1
````

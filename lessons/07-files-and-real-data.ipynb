## 2. Writing a file (~10 min)

To work with a file you **open** it. The friendliest way is a **`with` block**:

```python
with open("favorites.txt", "w") as f:
    f.write("hello")
```

Let's read that line by line:
- `open("favorites.txt", "w")` opens a file named `favorites.txt`. The `"w"` means **write mode** — *create the file (or wipe it clean) and let me write into it.*
- `as f` gives the open file a short name, `f`, so we can talk to it.
- The **indented** body (just like an `if` or a loop!) is where we do our writing. `f.write(...)` puts text into the file.

The magic of `with`: when the indented block ends, Python **closes the file for you automatically**. A closed file is a saved file — no extra step to remember. ✨

````python
# Write a single line into a brand-new file
with open("favorites.txt", "w") as f:
    f.write("Pizza is my favorite food.")

print("Done! I wrote favorites.txt 💾")
````

👀 **The delight moment:** look at the **file browser** on the left side of JupyterLab. There's `favorites.txt` — a real file, right next to this notebook. **Double-click it** to open it in a tab and see your text. You just made something that outlives the program! 🎉

### Writing multiple lines with `\n`

`f.write` does **not** add line breaks for you — if you call it twice, the text runs together. To start a new line you add `"\n"` yourself. `\n` is the **newline** character: a single invisible "press Enter" baked into the text.

````python
# Each "\n" ends a line and starts a fresh one
with open("favorites.txt", "w") as f:
    f.write("Pizza\n")
    f.write("Tacos\n")
    f.write("Sushi\n")

print("Wrote three lines. Re-open favorites.txt to see them stacked!")
````

### Adding on, instead of replacing: mode `"a"`

`"w"` always **wipes the file first**. When you want to *keep* what's there and tack new text onto the end — a diary, a log — use mode **`"a"`** for **append**.

In the cell below we first (re)write the file with `"w"` so it's the same every time you run this, *then* append a line with `"a"`. Doing the `"w"` part first is what makes this cell **safe to re-run** — otherwise every run would pile on another line.

````python
# Start fresh with "w" so re-running stays tidy...
with open("favorites.txt", "w") as f:
    f.write("Pizza\n")
    f.write("Tacos\n")

# ...then ADD a line on the end with "a" (append)
with open("favorites.txt", "a") as f:
    f.write("Ice cream\n")

print("favorites.txt now ends with the appended line.")
````

## 3. Reading it back (~14 min)

Writing is only half the story. To **read** a file, open it in mode **`"r"`** (read). The simplest move is `.read()`, which hands back the *entire* file as one big string.

````python
# Read the whole file at once
with open("favorites.txt", "r") as f:
    contents = f.read()

print(contents)
````

Reading everything at once is fine for small files. But usually you want to work with **one line at a time** — to number them, search them, or add them up. A file is loopable, so you can write `for line in f:` and it hands you the lines one by one. 🔁

````python
# Loop over the file, one line per pass
with open("favorites.txt", "r") as f:
    for line in f:
        print("Line:", line)
````

### The invisible-newline gotcha ✂️

Look closely at that output — there's an extra blank line between each one. That's because every line still carries its trailing `"\n"` (the newline we wrote earlier), and `print` adds *another* line break on top.

The fix is **`.strip()`** — a string method that returns a copy with whitespace (spaces, tabs, and newlines) trimmed off **both ends**. It's the cleanup tool you'll reach for on almost every line you read from a file.

````python
# .strip() trims the trailing "\n" (and any stray spaces) off each line
clean_lines = []
with open("favorites.txt", "r") as f:
    for line in f:
        clean = line.strip()      # tidy line, no newline on the end
        clean_lines.append(clean) # build a list (week 4 pattern!)

print(clean_lines)
````

Now `clean_lines` is a normal list of strings — exactly the kind you looped over in Lesson 4. Files in, lists out. 📋

## 4. Real data: comma-separated lines (~12 min)

Here's where files get *useful*. A huge amount of real data is stored as plain text with **one record per line** and **commas between the fields**, like:

```
milk,2.50
bread,3.20
eggs,4.00
```

This is exactly what a spreadsheet hands you when you choose *"Export as CSV"* (comma-separated values). Each line is a little record; the commas mark where one piece of data ends and the next begins.

The key tool is the string method **`.split(",")`**, which chops a string at every comma and gives back a **list** (week 4 again!).

````python
# .split(",") turns one line into a list of pieces
line = "milk,2.50"
parts = line.split(",")
print(parts)            # -> ['milk', '2.50']
print(parts[0])         # the name  -> milk
print(parts[1])         # the price -> 2.50  (still TEXT, not a number yet!)
````

Notice `parts[1]` is the **string** `"2.50"`, not the number `2.5`. To do math we must convert it with `float()` (a decimal number). But real data is messy — a stray blank or a typo will make `float()` raise a **`ValueError`** and crash the whole loop. 💥

**A handy shortcut:** when `.split(",")` gives back *exactly two* pieces, you can grab them into two names in one line — `name, price_text = parts` — instead of writing `parts[0]` and `parts[1]` yourself. That's called **unpacking**, and we'll use it from here on.

One more new face you'll spot in the output: **`{price:.2f}`** inside an f-string means *"show this number with 2 decimal places"* — a tidy formatting shortcut so prices line up as `2.50` instead of `2.5`.

This is where Lesson 6 pays off: we wrap the risky conversion in **`try` / `except`** so one bad line is *skipped* instead of stopping everything. Watch — the data below has one deliberately broken line.

````python
# Build a small shopping file: name,price per line (one line is messy on purpose)
with open("shopping.txt", "w") as f:
    f.write("milk,2.50\n")
    f.write("bread,3.20\n")
    f.write("eggs,oops\n")     # <-- broken: "oops" is not a number
    f.write("apples,4.00\n")

print("Wrote shopping.txt (with one messy line).")
````

````python
# Read it back, parse each line, total the prices — skipping bad lines
total = 0.0
with open("shopping.txt", "r") as f:
    for line in f:
        parts = line.strip().split(",")     # tidy, then split into pieces
        name, price_text = parts             # unpack the two pieces (see above)
        try:
            price = float(price_text)        # might raise ValueError
            total += price                   # accumulator (week 4)
            print(f"{name}: {price:.2f}")
        except ValueError:
            print(f"Skipping bad line for '{name}' (price was '{price_text}')")

print(f"\nTotal of the good items: ${total:.2f}")
````

See how the `eggs,oops` line was reported and skipped, while the total stayed correct? That's a small, robust data program — read, parse, convert safely, accumulate, report. 🛠️

## 5. Putting it together (~3 min)

One compact program that uses the whole pipeline: **write** a tiny scores file, **read** it back, **parse** each line, **compute** an average, and **report** it.

````python
# Write a small scores file: name,score
with open("scores.txt", "w") as f:
    f.write("Sam,72\n")
    f.write("Riya,90\n")
    f.write("Alex,85\n")

# Read, parse, and average the scores
total = 0
count = 0
with open("scores.txt", "r") as f:
    for line in f:
        parts = line.strip().split(",")
        name, score_text = parts
        total += int(score_text)   # whole-number scores
        count += 1

average = total / count
print(f"Class average over {count} students: {average:.1f}")
````

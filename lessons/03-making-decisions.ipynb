## 2. True or False — booleans (~8 min)

A **boolean** (named after mathematician George Boole) is a value that is either `True` or `False`. Those two words are the answers to yes/no questions, and Python uses them everywhere.

You get a boolean by **comparing** two things. A comparison asks a question and hands back `True` or `False`:

| You write | It asks |
|---|---|
| `a == b` | are they **equal**? |
| `a != b` | are they **not equal**? |
| `a < b`  | is `a` less than `b`? |
| `a <= b` | less than **or equal**? |
| `a > b`  | greater than? |
| `a >= b` | greater than **or equal**? |

````python
age = 20

print(age == 20)   # are they equal? -> True
print(age != 20)   # not equal?      -> False
print(age < 18)    # is age under 18?
print(age >= 21)   # 21 or older?
````

`True` and `False` are a real **type**, just like numbers and text. We can prove it with `type()`:

````python
result = 7 > 3
print(result)
print(type(result))   # <class 'bool'>  ('bool' is short for boolean)
````

### ⚠️ The `==` vs `=` trap (read this twice!)

This is the single most common beginner mistake, so we're calling it out **loudly**:

- **`=` means assign** — "put this value into this variable." `age = 20` *stores* 20.
- **`==` means ask** — "are these two equal?" `age == 20` *checks* and gives back `True`/`False`.

One equals sign **changes** a value. Two equals signs **compare** values. When you mean to ask a question, you almost always want **`==`**.

````python
score = 90          # one =  : assign 90 to score
print(score == 90)  # two == : ask "is score 90?"  -> True
print(score == 100) # ask "is score 100?"          -> False
````

## 3. if / elif / else (~14 min)

Now we use those `True`/`False` answers to **choose** what runs. The keyword is **`if`**:

```
if some_condition:
    do this when it's True
```

Two parts you must get right:

1. **The colon `:`** at the end of the `if` line. It says "the decision starts here."
2. **The indented block** below it. The lines that belong *inside* the `if` are pushed in (4 spaces).

➡️ **Indentation is how Python knows what belongs inside the `if`** — this is the single most important formatting rule in the whole language. Indented lines run only when the condition is `True`; lines back at the left margin always run.

````python
password_attempts = 3

if password_attempts > 0:
    print("You still have attempts left.")   # indented -> inside the if
    print("Try again carefully.")            # also inside

print("(this line is NOT indented, so it always runs)")
````

### Adding `else`

`else` gives you a "otherwise" path. Exactly one of the two blocks runs.

````python
hour = 20

if hour < 18:
    print("Good afternoon! ☀️")
else:
    print("Good evening! 🌙")
````

### `elif` for more than two choices

`elif` (short for "else if") lets you check more conditions in order. Python tries each one from the top and runs the **first** block whose condition is `True`, then skips the rest. `else` at the bottom catches everything that didn't match.

Here's a classic: a ticket price that depends on age.

````python
age = 15   # <- a preset variable; try changing it and re-running

if age < 5:
    price = 0       # toddlers ride free
elif age < 18:
    price = 8       # kids/teens
elif age < 65:
    price = 12      # adults
else:
    price = 6       # seniors

print(f"Ticket price: ${price}")
````

## 4. Combining conditions: and / or / not (~7 min)

Often a decision depends on **more than one** thing. Three little words combine booleans:

- **`and`** — `True` only if **both** sides are `True`.
- **`or`** — `True` if **at least one** side is `True`.
- **`not`** — flips a boolean: `not True` is `False`.

````python
print(True and False)   # both? -> False
print(True or False)    # either? -> True
print(not True)         # flip -> False

# These work on comparisons too:
print(7 > 3 and 2 > 5)  # one side is False -> False
````

Here's a realistic check: a shop gives **free shipping** if the order total is at least $50 **and** the shopper is a member.

````python
total = 60
is_member = True

if total >= 50 and is_member:
    print("Free shipping! 🚚")
else:
    print("Shipping costs $5.")
````

## 5. Repeating with while (~8 min) *(stretch goal — skip if short on time; come back to it any time)*

A **`while` loop** repeats a block **as long as** a condition stays `True`. It checks the condition, runs the block, checks again, runs again… and stops the moment the condition becomes `False`.

```
while condition:
    do this again and again
```

The golden rule: **something inside the loop must eventually make the condition `False`**, or the loop runs forever. Here we count down by subtracting 1 each time.

````python
count = 5

while count > 0:
    print(f"{count}...")
    count = count - 1     # this line is what eventually stops the loop

print("Lift-off! 🚀")
````

### ⚠️ Infinite loops — and how to escape one

If the condition never becomes `False`, the loop **never stops**. For example, if we forgot the `count = count - 1` line above, `count` would stay 5 forever.

If a cell seems stuck running (you'll see `[*]` beside it and it never finishes), don't panic:

- Click the **stop ■ button** in the toolbar, **or**
- Use the menu: **Kernel → Interrupt Kernel**.

That tells Python to stop the runaway cell so you can fix it and run again. Everyone hits this at least once — it's normal! 😅

### `break` — leave a loop early

`break` immediately exits the loop, even if the condition is still `True`. It's handy for stopping as soon as you've found what you wanted.

````python
number = 0

while number < 100:
    number = number + 1
    if number == 3:
        print("Found 3 — stopping early.")
        break          # jump out of the loop right now

print(f"Loop ended at number = {number}")
````

### Demo: the computer guesses your number 🔢 *(optional — a cool extra)*

**Don't worry about the logic here — just run it and watch.** 😎 This is a fun bonus to show what loops *can* do; nothing later in the course depends on you understanding how it works.

The computer plays a guessing game **against itself**. There's a fixed secret number, and each round it guesses the middle of its current range and shrinks the range based on whether the guess was too high or too low. Everything is fixed, so it runs the same way every time.

👉 The thing to notice: a `while` loop and an `if/elif/else` working together can solve a real problem. That's it — enjoy the show!

````python
secret = 42        # the number to find
low, high = 1, 100  # the range it could be in
guesses = 0

while low <= high:
    guess = (low + high) // 2   # // is whole-number division -> the middle
    guesses += 1
    print(f"Guess {guesses}: is it {guess}?")

    if guess == secret:
        print(f"🎉 Got it! The number was {secret} in {guesses} guesses.")
        break
    elif guess < secret:
        print("   too low — aim higher")
        low = guess + 1
    else:
        print("   too high — aim lower")
        high = guess - 1
````

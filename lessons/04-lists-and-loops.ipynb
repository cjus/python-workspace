## 2. Lists (~12 min)

A **list** is an ordered collection of values. You write one with **square brackets** `[ ]`, separating the items with commas. The items can be text, numbers, or a mix.

````python
# A list of strings (text). Note the square brackets and commas.
fruits = ["apple", "banana", "cherry"]
print(fruits)

# A list of numbers
scores = [88, 92, 75, 100]
print(scores)

# How many items? The built-in len() function (short for "length") counts them.
print("number of fruits:", len(fruits))
````

### Indexing — counting starts at 0 ⚠️

To grab one item, put its position — its **index** — in square brackets after the list.

Here's the classic surprise that trips up *everybody* at first: **Python counts from 0, not 1.** So the *first* item is at index `0`, the second is at index `1`, and so on.

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[0])   # FIRST item  -> apple
print(fruits[1])   # second item -> banana
print(fruits[2])   # third item  -> cherry
````

**Negative indexes** count from the *end*, which is handy for "the last one." `-1` is the last item, `-2` is the second-to-last.

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[-1])   # last item  -> cherry
print(fruits[-2])   # second-to-last -> banana
````

### Changing a list

Lists can be **changed** after you make them:

- **Replace** an item by assigning to its index: `my_list[1] = "new value"`
- **Add** an item to the end with `.append(...)`
- **Remove** an item by value with `.remove(...)`

And you can ask whether something is in a list with the `in` operator, which gives back `True` or `False`.

````python
fruits = ["apple", "banana", "cherry"]

fruits[1] = "blueberry"     # replace the item at index 1
fruits.append("date")       # add to the end
fruits.remove("apple")      # remove by value
print(fruits)

print("cherry" in fruits)   # is "cherry" in the list?
print("apple" in fruits)    # we removed it, so...?
````

## 3. The for loop (~12 min)

Suppose you want to print every fruit. You *could* write a `print` for each one... but that's tedious, and it breaks the moment the list changes length.

The **for loop** does it for you. It reads almost like English: *for each item in this list, do the following.*

```python
for item in my_list:
    # this indented part runs once per item
```

Just like the `if` blocks from last week, the body of the loop is **indented** (4 spaces). The indentation is how Python knows which lines belong to the loop. 🔁

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("I like", fruit)

print("Done!")   # not indented, so it runs once, after the loop
````

`fruit` is just a name *we* picked for "the current item." On each pass through the loop it holds the next value. You can call it anything, but a clear, singular name (`fruit`, `score`, `name`) makes the loop easy to read.

### Looping over a string

A string is a sequence of characters, so you can loop over it too — one character at a time.

````python
for letter in "cat":
    print(letter)
````

### `range()` — do this N times 🔢

Sometimes you don't have a list — you just want to repeat something a set number of times. `range(N)` gives you the numbers `0, 1, 2, ... up to but NOT including N`.

So `range(5)` produces `0, 1, 2, 3, 4` — that's five numbers, starting at 0 (there's that 0 again!).

````python
for i in range(5):
    print("Hello number", i)
````

`range` can also take a *start* and a *stop*: `range(1, 6)` gives `1, 2, 3, 4, 5` (starts at 1, stops before 6). Here's a times-table built with a loop:

````python
number = 7
for i in range(1, 6):        # i goes 1, 2, 3, 4, 5
    print(number, "x", i, "=", number * i)
````

### `range` + index — loop *with* the position

The plain `for fruit in fruits` loop hands you each item but not its **position**. When you also want the index ("item #1, item #2, ..."), loop over `range(len(my_list))` — that gives you each valid index `0, 1, 2, ...`, and `my_list[i]` reads the item there. 🔢

````python
fruits = ["apple", "banana", "cherry"]

for i in range(len(fruits)):   # i goes 0, 1, 2 — one index per item
    print(i, fruits[i])
````

## 4. Loop patterns you'll use forever (~12 min)

A handful of loop "shapes" come up again and again. Learn these four and you can solve a huge range of problems. 🛠️

### Pattern 1 — Add it all up (the *accumulator*)

Start a total at `0`, then add each item to it as you loop. `total += price` is shorthand for `total = total + price`.

The key idea: you set `total = 0` **before** the loop (once), and add **inside** the loop (once per item).

````python
prices = [3.50, 1.25, 8.00, 2.75]

total = 0            # start the running total BEFORE the loop
for price in prices:
    total += price   # add each price to the total
print("Total:", total)
````

### Pattern 2 — Count the matches

A *counter* is just an accumulator that adds `1` whenever something is true. Here: how many grades are 90 or above?

````python
grades = [88, 92, 75, 100, 90, 67]

count = 0
for grade in grades:
    if grade >= 90:      # only count the high ones
        count += 1
print("Grades 90 or above:", count)
````

### Pattern 3 — Find the biggest

Keep track of the best-so-far. Start it at the first item, then replace it whenever you see something bigger.

````python
scores = [42, 17, 88, 63, 91, 50]

biggest = scores[0]          # assume the first is the biggest, for now
for score in scores:
    if score > biggest:
        biggest = score      # found a new champion
print("Biggest score:", biggest)
````

### Pattern 4 — Build a new list

Start with an **empty list** `[]` and `.append()` to it inside the loop. Here we double every number.

````python
numbers = [1, 2, 3, 4]

doubled = []                 # an empty list to fill up
for n in numbers:
    doubled.append(n * 2)
print(doubled)
````

### Combine `for` + `if` — keep only some items

A small twist on Pattern 4: pair the loop with an `if` to **filter** as you build. This keeps only the even numbers (a number is even when the remainder after dividing by 2 is 0, written `n % 2 == 0`).

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

evens = []
for n in numbers:
    if n % 2 == 0:
        evens.append(n)
print("Even numbers:", evens)
````

## 5. Putting it together (~3 min)

Let's combine a few patterns: loop over a list of quiz scores, **count** how many are passing (60 or above), and report it with an **f-string** (from Lesson 2 — the `f"..."` that lets us drop variables right into the text with `{ }`).

````python
quiz_scores = [72, 45, 90, 58, 88, 61]

passes = 0
for score in quiz_scores:
    if score >= 60:
        passes += 1

print(f"{passes} out of {len(quiz_scores)} students passed. 🎉")
````

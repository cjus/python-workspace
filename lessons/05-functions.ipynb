## 2. Why functions? (~5 min)

Here's a secret: you've been *using* functions since Lesson 1. Every time you wrote `print(...)`, `len(...)`, or `range(...)`, you **called** a function — you asked a named piece of code to do its job.

A **function** is just a named recipe:
- **Define it once** — write down the steps and give them a name.
- **Use it many times** — "call" the name whenever you need those steps.

This keeps your code short, easy to read, and easy to fix (change the recipe in one place, and every use updates). Below, notice how we *call* a built-in function — that's the same thing you'll soon be writing yourself.

````python
# Calling functions that Python gives you for free
message = "hello functions"
print(len(message))   # len() hands back how many characters
print(message.upper())  # .upper() hands back a SHOUTING copy
````

## 3. Defining and calling (~10 min)

To make your own function you use the keyword **`def`** (short for *define*). Here is the anatomy:

```
def greet():        # def, a name, parentheses (), then a colon :
    print("Hi!")    # the indented body — the steps to run
```

- **`def`** tells Python "a function is coming."
- **`greet`** is the name you choose.
- **`()`** is where inputs go (empty for now).
- The **`:`** and the **indented body** are the recipe's steps.

Important: *defining* a function does **nothing** on its own — it just teaches Python the recipe. You have to **call** it (write its name with `()`) to actually run it.

````python
def greet():
    print("Hi there! 👋")

# Defining it above ran nothing. Now we CALL it — twice:
greet()
greet()
````

### Giving a function an input

The empty `()` can hold a **parameter** — a placeholder for a value you pass in when you call. Below, `name` is a parameter; when we call `greet("Sam")`, `name` becomes `"Sam"` inside the function.

````python
def greet(name):
    print(f"Hi there, {name}! 👋")

greet("Sam")
greet("Riya")
````

You can have **multiple parameters**, separated by commas. The values you pass line up in order.

````python
def greet_pair(name1, name2):
    print(f"Hi {name1} and {name2}! 👋")

greet_pair("Sam", "Riya")
````

## 4. `return` — the big one (~10 min)

This is the idea beginners most often mix up, so let's be crystal clear. 🧠

- **`print`** *shows* a value on the screen. That's it — once it's printed, your code can't reuse it.
- **`return`** *hands the value back* to wherever the function was called, so the **rest of your code can use it**.

Watch the difference. Here are two almost-identical functions.

````python
# Version A: PRINTS the result
def add_print(a, b):
    print(a + b)

# Version B: RETURNS the result
def add_return(a, b):
    return a + b

add_print(2, 3)        # shows 5 on screen
add_return(2, 3)       # hands 5 back... but we didn't catch it, so nothing shows
````

Notice version B looked like it did *nothing*. It actually handed `5` back — we just didn't catch it. The power of `return` is what you can do **afterwards**: store it, do more math, pass it on.

````python
# Catch the returned value in a variable, then keep using it:
total = add_return(2, 3)
print("total is", total)
print("double that is", total * 2)

# Try the same with the print version:
captured = add_print(10, 20)   # this prints 30...
print("captured is", captured) # ...but captured is None — nothing was handed back
````

See that? `add_print` gave back `None` (Python's word for "nothing"), because a function that doesn't `return` a value returns `None`. If you want to *use* a result later, you need `return`.

## 5. Polish: docstrings & defaults (~6 min)

Two small touches make your functions friendlier.

**Docstring** — a short text right under the `def` line, in triple quotes, describing what the function does. Python (and `help()`) can read it back to you.

````python
def area(width, height):
    """Return the area of a rectangle."""
    return width * height

help(area)   # Python reads your docstring back to you
````

**Default value** — give a parameter a fallback by writing `=value` in the `def` line. If the caller leaves it out, the default is used.

````python
def greet(name, greeting="Hello"):
    """Greet a person, defaulting to a plain Hello."""
    return f"{greeting}, {name}!"

print(greet("Sam"))                 # uses the default greeting
print(greet("Riya", "Welcome"))     # supplies its own greeting
````

## 6. Borrowing code: `import` (~7 min)

You don't have to write everything yourself. Python comes with a huge collection of ready-made functions grouped into **modules** — people call this the *"batteries included"* standard library. To use a module, you **`import`** it.

`math` is full of math helpers. Once imported, you reach inside it with a dot: `math.sqrt(...)`, `math.pi`.

````python
import math

print(math.sqrt(144))   # square root -> 12.0
print(math.pi)          # the value of pi
print(round(math.pi, 2))
````

The `random` module makes random values — great for games. We call **`random.seed(42)`** first so the "random" numbers come out the same every run (handy for a lesson). `random.randint(a, b)` gives a whole number between `a` and `b`, inclusive.

````python
import random

random.seed(42)   # makes the results repeatable

def roll_dice():
    """Return a random dice roll from 1 to 6."""
    return random.randint(1, 6)

print("You rolled:", roll_dice())
print("You rolled:", roll_dice())
print("You rolled:", roll_dice())
````

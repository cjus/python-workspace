## 2. What is Python? (~7 min)

**Python is a way to write instructions that a computer follows.** You type the steps in (almost) plain English, and the computer does exactly what you said — instantly, and as many times as you like.

Python is *everywhere* in the real world:

- 📱 The backends of apps like **Instagram** and **Spotify**
- 🔬 **NASA** and scientists use it to crunch data and run experiments
- 🤖 Most of today's **AI** is built with Python
- 🎮 Games, and 📊 automating boring **spreadsheet** work at jobs of every kind

It's also the classic *first* language because it reads so much like English. Here's a tiny real program — **don't worry about how it works yet**, just notice that you can almost *read* it:

````python
# Don't worry about how this works yet — just notice it reads like English.
for snack in ["apple", "cookie", "carrot"]:
    print("I would like a", snack, "please")
````

Could you guess what that does before running it? That readability is Python's superpower — and by the end of the course, lines like these will feel completely natural. ✨

### A first peek at variables 📦

Often we want to *remember* a value and reuse it. We give it a name — that's a **variable**. Here we store a name, then use it in a greeting:

````python
name = "Ada"
print(f"Hello, {name}! Welcome to Python.")
````

A few things just happened: `name = "Ada"` stored the text `"Ada"` under the name `name`. The `f"..."` text (an *f-string*) lets us drop that variable right inside the message using `{name}`. We'll **properly meet variables next week** — for now, just enjoy the magic.

Try it below: put your own name in the quotes and re-run. 👇

````python
# 🖊️ Your turn: change the name to yours, then re-run.
name = "friend"
print(f"Hello, {name}! Welcome to Python.")
````

### Fun one-liners 🎈

Python can do some delightfully silly (and powerful) things in a single line. Run this:

````python
print("🐍" * 10)
print(len("supercalifragilisticexpialidocious"))
print(2 ** 100)
````

Don't worry about *how* any of those work — just enjoy the results. (Ten snakes! A number bigger than your phone's calculator can show!) We'll explain bits like these in the weeks ahead. ✨

## 4. A taste of where you're heading (~8 min)

### Demo A: a chart from a few lines of code 📊

These few lines draw a real graph. Run it:

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
plants_grown = [1, 3, 6, 10, 15]

plt.plot(months, plants_grown, marker="o")
plt.title("My imaginary plant collection")
plt.ylabel("number of plants")
plt.show()
````

A handful of lines, and you have a chart you could drop into a report. Imagine doing that with your *own* data.

## 2. Dictionaries (~11 min)

A **dictionary** stores pairs: a **key** points to a **value**.

Think of a real paper dictionary: you look up a *word* (the key) to find its *definition* (the value). A list uses number positions (`scores[0]`); a dictionary lets you look things up by name instead.

You write one with curly braces `{ }`, putting `key: value` pairs inside, separated by commas.

````python
# A word -> definition dictionary
glossary = {
    "variable": "a name that stores a value",
    "loop": "code that repeats",
}

print(glossary)
````

### Reading a value: `dict[key]`

To get a value back out, put its **key** in square brackets — just like a list, but with the key instead of a number.

````python
glossary = {
    "variable": "a name that stores a value",
    "loop": "code that repeats",
}

print(glossary["variable"])   # look up the definition of "variable"
````

### Adding and updating

Assigning to a key that **doesn't exist yet** *adds* it. Assigning to a key that **already exists** *updates* it. Same syntax for both.

````python
glossary = {"variable": "a name that stores a value"}

glossary["function"] = "a reusable block of code"   # add a new pair
glossary["variable"] = "a name that holds a value"  # update an existing one

print(glossary)
````

### Checking keys with `in`, and the safe lookup `.get()`

Looking up a key that isn't there causes an error. Two ways to stay safe:

- `"key" in mydict` is `True`/`False` — handy in an `if` (week 3).
- `mydict.get("key", default)` returns the value if the key exists, or the `default` you give if it doesn't — no error.

````python
glossary = {"variable": "a name that stores a value"}

print("variable" in glossary)   # True
print("widget" in glossary)     # False

# .get() never crashes — it returns the fallback when the key is missing
print(glossary.get("variable", "not found"))
print(glossary.get("widget", "not found"))
````

### Looping with `.items()`

`for key, value in mydict.items():` walks through *every* pair at once, combining dictionaries with the `for` loop from week 4.

A dictionary is also a tidy way to **count** things: `counts.get(item, 0)` gives the count so far (0 if we haven't seen it yet), and we add 1 and store it back.

````python
glossary = {
    "variable": "a name that stores a value",
    "loop": "code that repeats",
}

# loop every pair
for word, definition in glossary.items():
    print(f"{word}: {definition}")

# the counting pattern: tally a few votes
votes = ["yes", "no", "yes", "yes"]
counts = {}
for vote in votes:
    counts[vote] = counts.get(vote, 0) + 1

print(counts)   # {'yes': 3, 'no': 1}
````

### Step (a): the questions — a list of dictionaries  *(weeks 4 + 8)*

Each question is a **dictionary** with a `"question"` and an `"answer"`. We put several of them in a **list**. A list of dictionaries is an everyday way to store structured data.

````python
questions = [
    {"question": "What keyword starts a function in Python?", "answer": "def"},
    {"question": "What do we call a name that stores a value?", "answer": "variable"},
    {"question": "What punctuation ends an if-statement line?", "answer": ":"},
    {"question": "Which loop repeats a fixed number of times: for or while?", "answer": "for"},
]

print(f"Loaded {len(questions)} questions.")
print(questions[0]["question"])   # peek at the first question
````

### Step (b): load the questions from a **file**  *(files = week 7, try/except = week 6)*

Real programs keep their data *outside* the code, so you can edit the questions without touching the program. Let's do that. First we **write** a small `quiz_questions.txt` file with one `question,answer` per line (so the notebook stays self-contained and safe to re-run). Then a `load_questions` function **reads** it back into the same list-of-dictionaries shape.

We wrap the reading in `try` / `except FileNotFoundError` (week 6) so that if the file is missing, we get a friendly message and an empty list instead of a crash.

````python
# Write the questions file (mode "w" recreates it cleanly every run) — week 7
with open("quiz_questions.txt", "w") as f:
    f.write("What keyword starts a function in Python?,def\n")
    f.write("What do we call a name that stores a value?,variable\n")
    f.write("What punctuation ends an if-statement line?,:\n")
    f.write("Which loop repeats a fixed number of times: for or while?,for\n")

print("Wrote quiz_questions.txt ✅")
````

````python
def load_questions(filename):
    """Read 'question,answer' lines from a file into a list of dictionaries."""
    loaded = []
    try:
        with open(filename) as f:                 # open the file (week 7)
            for line in f:
                line = line.strip()               # trim the newline (week 7)
                question, answer = line.split(",")  # split on the comma (week 7)
                loaded.append({"question": question, "answer": answer})
    except FileNotFoundError:                     # handle a missing file (week 6)
        print(f"⚠️  Couldn't find {filename} — starting with no questions.")
        return []
    return loaded


questions = load_questions("quiz_questions.txt")
print(f"Loaded {len(questions)} questions from the file.")
print(questions[0]["question"])   # same shape as before — the rest of the game just works
````

### Step (c): a checker function  *(weeks 2 + 3 + 5)*

We write a **function** (week 5) that compares the given answer to the expected one. To be fair, we make it **case-insensitive** with `.lower()` (text handling from week 2) and use a **comparison** that returns `True`/`False` (week 3). We also tidy up with `.strip()` — you met it in week 7; here it trims any accidental spaces off the **ends** of the text, so `" def "` still counts as `"def"`.

````python
def check_answer(given, expected):
    """Return True if `given` matches `expected`, ignoring case and spaces."""
    return given.strip().lower() == expected.strip().lower()


# quick tests
print(check_answer("DEF", "def"))     # True  -> case doesn't matter
print(check_answer(" def ", "def"))   # True  -> spaces trimmed
print(check_answer("class", "def"))   # False
````

### Step (d): the `play` function  *(weeks 2-7, all together!)*

This is the heart of the game. It loops through the questions (week 4), checks each answer with our function (week 5), prints right/wrong with f-strings (week 2), and **accumulates a score** in a variable (week 2) using an `if`/`else` (week 3).

We loop with `for i in range(len(questions)):` so `i` counts `0, 1, 2, …` — exactly the `range(len(...))` trick from Lesson 4's after-class notes. That gives us a number we can use to grab the matching question *and* the matching answer: `questions[i]` and `given_answers[i]`.

It takes a `given_answers` list — the answers to use — so the whole thing can run on its own, no typing required.

````python
def play(questions, given_answers):
    """Play through the quiz using a pre-scripted list of answers. Return the score."""
    score = 0
    for i in range(len(questions)):          # i goes 0, 1, 2, ... (range from week 4)
        item = questions[i]                  # the i-th question dictionary
        given = given_answers[i]             # the matching pre-set answer
        print(f"Q{i + 1}: {item['question']}")
        print(f"   You answered: {given}")
        if check_answer(given, item["answer"]):
            print("   ✅ Correct!")
            score = score + 1
        else:
            print(f"   ❌ Nope — the answer was: {item['answer']}")
        print()
    print(f"Final score: {score} out of {len(questions)}")
    return score
````

### Step (e): run it!  *(deterministic — same result every time)*

Here are some pre-set answers (a couple right, a couple wrong on purpose) so the quiz runs unattended.

````python
preset_answers = ["def", "variable", ";", "for"]   # the 3rd one is wrong on purpose

my_score = play(questions, preset_answers)
````
